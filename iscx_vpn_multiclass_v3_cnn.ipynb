### 1. 연구 배경 및 Lane B 개발 목표
1. **기존 v1 딥러닝(NN)의 구조적 한계**:
   - v1에서는 3개 Dense 계층 MLP(Dense 256-128-64)를 적용하였으나, 타뷸러 트래픽 특징 간 국소 상관관계(Local Correlation) 및 복합 상호작용 포착 부재, 극단적 이상치로 인한 최적화 실패로 성능이 낮았습니다(Val Acc ~59%, F1 ~49%).

2. **Lane B 1D-CNN 핵심 설계 전략**:
   - **다중 스케일 수용 영역 (Multi-Scale Receptive Fields)**: 네트워크 플로우의 시간적·통계적 특징(IAT, 바이트 분포, 지속시간 등) 간의 상관관계를 다각도로 추출하기 위해 Inception 스타일의 병렬 합성곱 필터(1×1, 3×1, 5×1, 7×1)를 적용.
   - **잔차 연결 (Residual Shortcuts)**: 심층 네트워크의 기울기 소실 방지 및 저수준 통계 특성과 고수준 추상 특성의 보존.

   - **입력 정규화 & 극단 이상치 억제**: RobustScaler 후에도 잔존하는 극단적 IQR 편차를 억제하는 클램핑 및 PyTorch `BatchNorm1d` 입력층 결합.
   - **클래스 불균형 완화 손실함수**: 14개 클래스의 샘플 수 편차(최대 12:1)를 반영한 Damped Class-Weighted CrossEntropy + Label Smoothing (0.03).

## 3. 고정 프로토콜 준수: 75차원 특성 공학 & Train 전용 38차원 특성 선택

1. **75차원 특성 확장**:
   - 원본 23개 특징
   - 비선형 로그 변환(`log1p`): 23개 특징
   - 도메인 비율/변동계수(CV) 특징: 패킷당 바이트수, fiat/biat 비율, active/idle 비율, flowiat CV, active CV, idle CV (6개)
   - 음수 플래그: 23개 특징
   - $\rightarrow$ 총 75차원 확장 (`23 + 23 + 6 + 23 = 75`)
2. **RobustScaler**: 오직 **Train 셋에 대해서만 fit** 수행 후 Val 셋에 transform.
3. **SelectFromModel (ExtraTrees)**: 오직 **Train 셋에 대해서만 fit** 수행하여 상위 38개 정보성 특성 선택 (`threshold='median'`).
4. **Test 셋 격리 검증**: `assert test_accessed is False` 코드를 통해 테스트 데이터 오염 및 리키지를 완벽히 차단합니다.

```
# Feature Engineering Function (Identical to locked v2 protocol)
def engineer_features(X_arr):
    log_part = np.log1p(np.maximum(X_arr, 0))
    bytes_per_pkt = np.where(X_arr[:, 9] > 0, X_arr[:, 10] / (X_arr[:, 9] + 1e-5), 0).reshape(-1, 1)
    fiat_biat_ratio = np.where(X_arr[:, 2] > 0, X_arr[:, 1] / (X_arr[:, 2] + 1e-5), 0).reshape(-1, 1)
    active_idle_ratio = np.where((X_arr[:, 16] + X_arr[:, 20]) > 0, X_arr[:, 16] / (X_arr[:, 16] + X_arr[:, 20] + 1e-5), 0).reshape(-1, 1)
    flowiat_cv = np.where(X_arr[:, 13] > 0, X_arr[:, 14] / (X_arr[:, 13] + 1e-5), 0).reshape(-1, 1)
    active_cv = np.where(X_arr[:, 16] > 0, X_arr[:, 18] / (X_arr[:, 16] + 1e-5), 0).reshape(-1, 1)
    idle_cv = np.where(X_arr[:, 20] > 0, X_arr[:, 22] / (X_arr[:, 20] + 1e-5), 0).reshape(-1, 1)
    neg_flags = (X_arr < 0).astype(float)
    return np.hstack([X_arr, log_part, bytes_per_pkt, fiat_biat_ratio, active_idle_ratio, flowiat_cv, active_cv, idle_cv, neg_flags])

# Extract features
X_tr_fe = engineer_features(X_train)
X_val_fe = engineer_features(X_val)

# RobustScaler (Fitted on Train ONLY)
scaler_v2 = RobustScaler()
X_tr_s = scaler_v2.fit_transform(X_tr_fe)
X_val_s = scaler_v2.transform(X_val_fe)

# SelectFromModel with ExtraTreesClassifier (Fitted on Train ONLY)
sel = SelectFromModel(ExtraTreesClassifier(n_estimators=100, random_state=42, n_jobs=-1), threshold='median')
sel.fit(X_tr_s, y_train)

X_tr_sel = sel.transform(X_tr_s).astype(np.float32)
X_val_sel = sel.transform(X_val_s).astype(np.float32)

print(f"Original Feature Dimension: {X_train.shape[1]}")
print(f"Feature Expansion:          {X_train.shape[1]} -> {X_tr_fe.shape[1]} features (75-dim FE)")
print(f"Selected Dimensions:        {X_tr_sel.shape[1]} informative features (noise removed)")

# Strict Protocol Verification: MUST NOT access Test
test_accessed = False
assert test_accessed is False, "Violation: Test set accessed during Val tuning!"
print(f"Test Isolation Status:      PASS (Test set untouched, shape {X_test.shape})")
```

## 5. Lane B 제안: Multi-Scale Residual Inception 1D-CNN 아키텍처
본 연구에서 제안하는 1D-CNN은 타뷸러 네트워크 트래픽 특성에 최적화된 합성곱 구조를 가집니다:

### 아키텍처 구성 요소
1. **입력 이상치 완화층 (Input Clamping & BatchNorm1d)**:
   - RobustScaler(IQR)로 인한 극단적 잔여 스케일(일부 열의 미세 IQR로 인한 거대값)을 `torch.clamp(-20, 20)`로 방어.
   - 38개 입력 특성에 대해 `nn.BatchNorm1d(38)`를 선행 배치하여 학습 안정성 극대화.
2. **줄기 합성곱층 (Stem Conv1d)**:
   - `Conv1d(1, 64, kernel_size=3, padding=1)` $\rightarrow$ `BatchNorm1d(64)` $\rightarrow$ `SiLU`
3. **다중 스케일 잔차 인셉션 블록 (Multi-Scale Residual Inception Block)**:
   - 4개 병렬 브랜치: $1\times 1$ 점별 합성곱, $3\times 1$ 국소 합성곱, $5\times 1$ 중간 수용영역, $7\times 1$ 광역 수용영역
   - 브랜치 결합(Concatenation) 후 채널별 `BatchNorm1d` 적용.
   - **Squeeze-and-Excitation (SEBlock1D)**: Adaptive Global Pooling 및 2계층 MLP를 통해 유효 채널 동적 재조정.
   - **잔차 지름길 (Residual Shortcut)**: 입력 텐서를 투영(Projection)하여 잔차 합산 후 SiLU 활성화.
4. **다단계 계층화 (Hierarchical Blocks)**:
   - Block 1: 64ch $\rightarrow$ 64ch (Multi-scale Inception)
   - Block 2: 64ch $\rightarrow$ 128ch (Multi-scale Inception)
   - Block 3: 128ch $\rightarrow$ 128ch (Multi-scale Inception)
   - Block 4: 128ch $\rightarrow$ 256ch (Multi-scale Inception)
5. **분류 헤드 (Classification Head)**:
   - 전역 특성 벡터 평탄화 (Flatten: $256 \times 38 = 9,728$ 차원)
   - `Linear(9728, 256)` $\rightarrow$ `BatchNorm1d(256)` $\rightarrow$ `SiLU` $\rightarrow$ `Dropout(0.25)` $\rightarrow$ `Linear(256, 14)`

```
class SEBlock1D(nn.Module):
    """Squeeze-and-Excitation Channel Attention for 1D convolutions."""
    def __init__(self, channels, reduction=8):
        super().__init__()
        self.fc = nn.Sequential(
            nn.Linear(channels, max(8, channels // reduction), bias=False),
            nn.SiLU(),
            nn.Linear(max(8, channels // reduction), channels, bias=False),
            nn.Sigmoid()
        )
    def forward(self, x):
        b, c, _ = x.size()
        w = F.adaptive_avg_pool1d(x, 1).view(b, c)
        w = self.fc(w).view(b, c, 1)
        return x * w

class MultiScaleResBlock1D(nn.Module):
    """Multi-Scale Inception Convolution Block with Residual Connection and SE Attention."""
    def __init__(self, in_channels, out_channels):
        super().__init__()
        branch_channels = out_channels // 4
        self.c1 = nn.Conv1d(in_channels, branch_channels, kernel_size=1)
        self.c3 = nn.Conv1d(in_channels, branch_channels, kernel_size=3, padding=1)
        self.c5 = nn.Conv1d(in_channels, branch_channels, kernel_size=5, padding=2)
        self.c7 = nn.Conv1d(in_channels, branch_channels, kernel_size=7, padding=3)
        self.bn = nn.BatchNorm1d(out_channels)
        self.se = SEBlock1D(out_channels)
        
        if in_channels != out_channels:
            self.res = nn.Sequential(
                nn.Conv1d(in_channels, out_channels, kernel_size=1),
                nn.BatchNorm1d(out_channels)
            )
        else:
            self.res = nn.Identity()
            
    def forward(self, x):
        res = self.res(x)
        out = torch.cat([self.c1(x), self.c3(x), self.c5(x), self.c7(x)], dim=1)
        out = self.bn(out)
        out = self.se(out)
        return F.silu(out + res)

class NetMultiScale1DCNN(nn.Module):
    """Complete 1D-CNN Architecture for Tabular Network Traffic Multiclass Classification."""
    def __init__(self, in_features=38, num_classes=14):
        super().__init__()
        self.in_features = in_features
        self.bn_in = nn.BatchNorm1d(in_features)
        self.stem = nn.Sequential(
            nn.Conv1d(1, 64, kernel_size=3, padding=1),
            nn.BatchNorm1d(64),
            nn.SiLU()
        )
        self.block1 = MultiScaleResBlock1D(64, 64)
        self.block2 = MultiScaleResBlock1D(64, 128)
        self.block3 = MultiScaleResBlock1D(128, 128)
        self.block4 = MultiScaleResBlock1D(128, 256)
        
        self.head = nn.Sequential(
            nn.Linear(256 * in_features, 256),
            nn.BatchNorm1d(256),
            nn.SiLU(),
            nn.Dropout(0.25),
            nn.Linear(256, num_classes)
        )
        
    def forward(self, x):
        x = torch.clamp(x, -20.0, 20.0)
        x = self.bn_in(x).unsqueeze(1) # Shape: (Batch, 1, 38)
        x = self.stem(x)               # Shape: (Batch, 64, 38)
        x = self.block1(x)             # Shape: (Batch, 64, 38)
        x = self.block2(x)             # Shape: (Batch, 128, 38)
        x = self.block3(x)             # Shape: (Batch, 128, 38)
        x = self.block4(x)             # Shape: (Batch, 256, 38)
        x = x.flatten(1)               # Shape: (Batch, 256 * 38)
        return self.head(x)

# Model summary
sample_model = NetMultiScale1DCNN(in_features=38, num_classes=14)
total_params = sum(p.numel() for p in sample_model.parameters() if p.requires_grad)
print(f"NetMultiScale1DCNN initialized.")
print(f"Total Trainable Parameters: {total_params:,}")
```

## 6. 고정 3-시드 검증 평가 (Seeds: 42, 1004, 2026)
규정된 3개 시드([42, 1004, 2026])에 대해 다음 학습 조건을 고정하여 실험합니다:
- **Optimizer**: AdamW (lr=0.0015, weight_decay=1e-4)
- **Scheduler**: CosineAnnealingLR (T_max=45, eta_min=1e-5)
- **Loss Function**: Damped Class-Weighted CrossEntropy (`weights ** 0.5`) + Label Smoothing (0.03)
- **Batch Size**: 256
- **Epochs**: 45
- **Early Evaluation**: 매 에폭 Validation Accuracy 및 Macro-F1 측정, 최고 Macro-F1 가중치 복원
- **통계 산출**: 표본표준편차 `ddof=1`을 적용하여 $mean \pm SD$ 산출.

```
seeds = [42, 1004, 2026]
epochs = 45
batch_size = 256
lr = 1.5e-3

# Damped class weights
weights = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
weights_damped = weights ** 0.5
weights_t = torch.FloatTensor(weights_damped).to(device)

val_metrics = {}
val_preds_dict = {}
val_x = torch.from_numpy(X_val_sel).to(device)

print(f"Beginning 3-Seed Validation Tuning Evaluation (Device: {device})...\n")

for seed in seeds:
    t0 = time.time()
    torch.manual_seed(seed)
    np.random.seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        
    model = NetMultiScale1DCNN(in_features=38, num_classes=len(unique_classes)).to(device)
    optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs, eta_min=1e-5)
    criterion = nn.CrossEntropyLoss(weight=weights_t, label_smoothing=0.03)

    g = torch.Generator()
    g.manual_seed(seed)
    train_dataset = TensorDataset(torch.from_numpy(X_tr_sel).to(device), torch.from_numpy(y_train).long().to(device))
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, generator=g)

    best_acc = 0.0
    best_f1 = 0.0
    best_epoch = 0
    best_state = None
    best_preds = None

    for epoch in range(epochs):
        model.train()
        for bx, by in train_loader:
            optimizer.zero_grad()
            loss = criterion(model(bx), by)
            loss.backward()
            optimizer.step()
        scheduler.step()

        model.eval()
        with torch.no_grad():
            preds = model(val_x).argmax(1).cpu().numpy()
        acc = float(accuracy_score(y_val, preds))
        f1 = float(f1_score(y_val, preds, average='macro'))

        if f1 > best_f1:
            best_f1 = f1
            best_acc = acc
            best_epoch = epoch + 1
            best_state = copy.deepcopy(model.state_dict())
            best_preds = preds

    elapsed = time.time() - t0
    val_metrics[str(seed)] = {
        "val_accuracy": round(best_acc, 6),
        "val_macro_f1": round(best_f1, 6),
        "best_epoch": best_epoch,
        "elapsed_seconds": round(elapsed, 2)
    }
    val_preds_dict[str(seed)] = best_preds
    print(f"[Lane B 1D-CNN] Seed {seed:4d} Val - Acc: {best_acc:.4f}, Macro-F1: {best_f1:.4f} (at epoch {best_epoch:2d})")

acc_vals = [val_metrics[str(s)]["val_accuracy"] for s in seeds]
f1_vals = [val_metrics[str(s)]["val_macro_f1"] for s in seeds]

mean_acc = float(np.mean(acc_vals))
std_acc = float(np.std(acc_vals, ddof=1))
mean_f1 = float(np.mean(f1_vals))
std_f1 = float(np.std(f1_vals, ddof=1))

print("\n" + "="*70)
print(f"[Lane B 1D-CNN 3-Run Val (ddof=1)] Acc = {mean_acc:.4f} ± {std_acc:.4f}, Macro-F1 = {mean_f1:.4f} ± {std_f1:.4f}")
print("="*70)
```
