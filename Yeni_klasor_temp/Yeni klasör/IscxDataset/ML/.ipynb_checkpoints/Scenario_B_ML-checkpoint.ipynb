```python
import pandas as pd
import lightgbm as lgb
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.io.arff import loadarff
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
raw_data = loadarff('../ScenarioB/TimeBasedFeatures-Dataset-15s.arff')
```

```python
df_B = pd.DataFrame(raw_data[0])
df_B
```

```python
df_B['class1'].value_counts()
```

```python
df_B['class1'] = df_B['class1'].replace([b'VOIP'],1).replace([b'VPN-BROWSING'],2).replace([b'BROWSING'],3).replace(
[b'VPN-VOIP'],4).replace([b'VPN-FT'],5).replace([b'VPN-CHAT'],6).replace([b'FT'],7).replace([b'P2P'],8).replace(
    [b'VPN-P2P'],9).replace([b'CHAT'],10).replace([b'VPN-MAIL'],11).replace([b'STREAMING'],12).replace(
    [b'VPN-STREAMING'],13).replace([b'MAIL'],14)
```

```python
X = df_B.drop(columns=["class1"])
y = df_B["class1"]
random_state = 1
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=True, random_state=random_state, stratify=y)
```

```python
randomForest = RandomForestClassifier(max_depth=20, random_state=random_state)
decisionTree = DecisionTreeClassifier(random_state = random_state)
lightGbm = lgb.LGBMClassifier(random_state=random_state)
```

```python
randomForest.fit(X_train, y_train)
print("Random Forest fitted")
decisionTree.fit(X_train, y_train)
print("Decision Tree fitted")
lightGbm.fit(X_train, y_train)
print("LightGBM fitted")
```

```python
randomForest_y_pred = randomForest.predict(X_test)
decisionTree_y_pred = decisionTree.predict(X_test)
lightGbm_y_pred = lightGbm.predict(X_test)
```

```python
randomForestResult = accuracy_score(randomForest_y_pred, y_test)
decisionTreeResult = accuracy_score(decisionTree_y_pred, y_test)
lightGBMResult = accuracy_score(lightGbm_y_pred, y_test)
```

```python
print('Scenario B Random Forest Accuracy Score: {0:0.4f}'.format(randomForestResult))
print('Scenario B Decision Tree Accuracy Score: {0:0.4f}'.format(decisionTreeResult))
print('Scenario B LightGBM Accuracy Score: {0:0.4f}'.format(lightGBMResult))
```

```python
randomForest_cm = confusion_matrix(y_test, randomForest_y_pred)
decisionTree_cm = confusion_matrix(y_test, decisionTree_y_pred)
lightGbm_cm = confusion_matrix(y_test, lightGbm_y_pred)
```

```python
target_names = ["b'VOIP'","b'VPN-BROWSING'","b'BROWSING'","b'VPN-VOIP'","b'VPN-FT'","b'VPN-CHAT'","b'FT'",
           "b'P2P'","b'VPN-P2P'","b'CHAT'","b'VPN-MAIL'","b'STREAMING'","b'VPN-STREAMING'","b'MAIL'"] 
randomForest_cm_df = pd.DataFrame(randomForest_cm,
                     index = target_names, 
                     columns = target_names)
decisionTree_cm_df = pd.DataFrame(decisionTree_cm,
                     index = target_names, 
                     columns = target_names)
lightGbm_cm_df = pd.DataFrame(lightGbm_cm,
                     index = target_names, 
                     columns = target_names)
```

```python
def createConfusionMatrix(matrix,title):
    plt.figure(figsize=(15,15))
    sns.heatmap(matrix, annot=True, fmt='g')
    plt.title(title, fontsize=20)
    plt.ylabel('Actual Values')
    plt.xlabel('Predicted Values')
    plt.show()
```

```python
randomForest_title = "Scenario B Random Forest Model Confusion Matrix"
createConfusionMatrix(randomForest_cm_df, randomForest_title)
```

```python
decisionTree_title = " Scenario B Decision Tree Model Confusion Matrix"
createConfusionMatrix(decisionTree_cm_df, decisionTree_title)
```

```python
lightGbm_title = "Scenario B LightGBM Model Confusion Matrix"
createConfusionMatrix(lightGbm_cm_df, lightGbm_title)
```

```python
randomForest_report = classification_report(y_test, randomForest_y_pred, target_names=target_names)
decisionTree_report = classification_report(y_test, decisionTree_y_pred, target_names=target_names)
lightGbm_report = classification_report(y_test, lightGbm_y_pred, target_names=target_names)
```

```python
print("     Scenario B Random Forest Classification Report\n", randomForest_report)
```

```python
print("     Scenario B Decision Tree Classification Report\n", decisionTree_report)
```

```python
print("       Scenario B LightGBM Classification Report\n", lightGbm_report)
```
