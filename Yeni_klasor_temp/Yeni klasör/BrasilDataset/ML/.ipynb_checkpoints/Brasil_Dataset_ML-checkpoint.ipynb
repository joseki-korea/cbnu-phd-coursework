```python
import pandas as pd
import numpy as np
import lightgbm as lgb
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.io.arff import loadarff
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
raw_data = loadarff('../Dataset/Day1.TCP.arff')
```

```python
df_Brasil = pd.DataFrame(raw_data[0])
df_Brasil
```

```python
df_Brasil['249'].value_counts()
```

```python
df_Brasil = df_Brasil[df_Brasil['249'] != b'GAMES']
df_Brasil = df_Brasil[df_Brasil['249'] != b'INTERACTIVE']
df_Brasil = df_Brasil[df_Brasil['249'] != b'MULTIMEDIA']
```

```python
df_Brasil['249'].value_counts()
```

```python
df_Brasil['249'] = df_Brasil['249'].replace([b'WWW'],1).replace([b'MAIL'],2).replace([b'BULK'],3).replace(
[b'DATABASE'],4).replace([b'P2P'],5).replace([b'SERVICES'],6).replace([b'ATTACK'],7).replace([b'MULTIMEDIA'],8).replace(
    [b'INTERACTIVE'],9)
```

```python
df_Brasil[df_Brasil.select_dtypes(np.float64).columns] = df_Brasil.select_dtypes(np.float64).astype(np.float32)
df_Brasil = df_Brasil.dropna()
df_Brasil.info()
```

```python
X = df_Brasil.drop(columns=["249"])
y = df_Brasil["249"]
random_state = 1
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=True, random_state=random_state, stratify=y)
```

```python
randomForest = RandomForestClassifier(max_depth=20, random_state=random_state)
decisionTree = DecisionTreeClassifier(random_state = random_state)
lightGbm = lgb.LGBMClassifier(random_state=random_state)
```

```python
randomForest.fit(X_train, y_train)
print("Random Forest fitted")
decisionTree.fit(X_train, y_train)
print("Decision Tree fitted")
lightGbm.fit(X_train, y_train)
print("LightGBM fitted")
```

```python
randomForest_y_pred = randomForest.predict(X_test)
decisionTree_y_pred = decisionTree.predict(X_test)
lightGbm_y_pred = lightGbm.predict(X_test)
```

```python
randomForestResult = accuracy_score(randomForest_y_pred, y_test)
decisionTreeResult = accuracy_score(decisionTree_y_pred, y_test)
lightGBMResult = accuracy_score(lightGbm_y_pred, y_test)
```

```python
print('Brasil Dataset Random Forest Accuracy Score: {0:0.4f}'.format(randomForestResult))
print('Brasil Dataset Decision Tree Accuracy Score: {0:0.4f}'.format(decisionTreeResult))
print('Brasil Dataset LightGBM Accuracy Score: {0:0.4f}'.format(lightGBMResult))
```

```python
randomForest_cm = confusion_matrix(y_test, randomForest_y_pred)
decisionTree_cm = confusion_matrix(y_test, decisionTree_y_pred)
lightGbm_cm = confusion_matrix(y_test, lightGbm_y_pred)
```

```python
target_names = ["b'WWW'","b'MAIL'","b'BULK'","b'DATABASE'","b'P2P'","b'SERVICES'","b'ATTACK'"] 
randomForest_cm_df = pd.DataFrame(randomForest_cm,
                     index = target_names, 
                     columns = target_names)
decisionTree_cm_df = pd.DataFrame(decisionTree_cm,
                     index = target_names, 
                     columns = target_names)
lightGbm_cm_df = pd.DataFrame(lightGbm_cm,
                     index = target_names, 
                     columns = target_names)
```

```python
def createConfusionMatrix(matrix,title):
    plt.figure(figsize=(10,10))
    sns.heatmap(matrix, annot=True, fmt='g')
    plt.title(title, fontsize=20)
    plt.ylabel('Actual Values')
    plt.xlabel('Predicted Values')
    plt.show()
```

```python
randomForest_title = "Brasil Dataset Random Forest Model Confusion Matrix"
createConfusionMatrix(randomForest_cm_df, randomForest_title)
```

```python
decisionTree_title = " Brasil Dataset Decision Tree Model Confusion Matrix"
createConfusionMatrix(decisionTree_cm_df, decisionTree_title)
```

```python
lightGbm_title = "Brasil Dataset LightGBM Model Confusion Matrix"
createConfusionMatrix(lightGbm_cm_df, lightGbm_title)
```

```python
randomForest_report = classification_report(y_test, randomForest_y_pred, target_names=target_names)
decisionTree_report = classification_report(y_test, decisionTree_y_pred, target_names=target_names)
lightGbm_report = classification_report(y_test, lightGbm_y_pred, target_names=target_names)
```

```python
print("     Brasil Dataset Random Forest Classification Report\n", randomForest_report)
```

```python
print("     Brasil Dataset Decision Tree Classification Report\n", decisionTree_report)
```

```python
print("       Brasil Dataset LightGBM Classification Report\n", lightGbm_report)
```
